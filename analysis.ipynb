* **Income_Category** - the customer's income category.

* **Months_on_book** - the number of months the customer has been with the bank.

* **Total_Relationship_Count** - the total number of products or services the customer has with the bank.

* **Months_Inactive_12_mon** - the number of months during the previous 12 months when the customer was inactive.

* **Contacts_Count_12_mon** - the number of times the customer contacted the bank during the previous 12 months.

* **Credit_Limit** - the credit limit assigned to the customer's credit card.

* **Total_Revolving_Bal** - the total revolving balance on the customer's credit card.

* **Avg_Utilization_Ratio** - the average percentage of the available credit that the customer used.

* **Total_Trans_Amt** - the total transaction amount made by the customer during the observed period.

* **Total_Trans_Ct** - the total number of transactions made by the customer.

* **Total_Ct_Chng_Q4_Q1** - the change in the number of transactions between two periods.

* **Total_Amt_Chng_Q4_Q1** - the change in the total transaction amount between two periods.

## Section 7: Answers to the Exploratory Questions


### Question 1: Is customer age related to credit card transaction activity?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
if 'df' not in globals():
    df = pd.read_csv('BankChurners.csv')

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
sns.regplot(data=df, x='Customer_Age', y='Total_Trans_Amt', scatter_kws={'alpha': 0.18, 's': 14}, line_kws={'color': 'crimson'}, ax=axes[0])
axes[0].set_title('Age vs. total transaction amount')
axes[0].set_xlabel('Customer age')
axes[0].set_ylabel('Total transaction amount')
sns.regplot(data=df, x='Customer_Age', y='Total_Trans_Ct', scatter_kws={'alpha': 0.18, 's': 14}, line_kws={'color': 'crimson'}, ax=axes[1])
axes[1].set_title('Age vs. total transaction count')
axes[1].set_xlabel('Customer age')
axes[1].set_ylabel('Total transaction count')
fig.suptitle('Question 1: Age has a negligible relationship with transaction activity', y=1.02)
plt.tight_layout()
plt.show()
```

**Answer and analysis:** Customer age has almost no linear relationship with transaction activity. The Pearson correlations are approximately **-0.046** for total transaction amount and **-0.067** for transaction count. Both regression lines are nearly flat and the points are widely dispersed. Therefore, older customers make only slightly fewer transactions on average, but the relationship is too weak to be practically important in this dataset. The expectation of a weak relationship is supported.

### Question 2: Are there substantial differences in customer attrition between income categories?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
if 'df' not in globals():
    df = pd.read_csv('BankChurners.csv')

income_order = ['Less than $40K', '$40K - $60K', '$60K - $80K', '$80K - $120K', '$120K +', 'Unknown']
income_rates = (df.assign(Attrited=df['Attrition_Flag'].eq('Attrited Customer'))
                .groupby('Income_Category', observed=False)['Attrited'].mean().reindex(income_order).mul(100).reset_index())
plt.figure(figsize=(10, 5))
ax = sns.barplot(data=income_rates, x='Income_Category', y='Attrited', color='#4c78a8')
for i, value in enumerate(income_rates['Attrited']):
    ax.text(i, value + 0.35, f'{value:.1f}%', ha='center', fontsize=9)
plt.title('Question 2: Attrition rates by income category')
plt.xlabel('Income category')
plt.ylabel('Attrition rate (%)')
plt.xticks(rotation=25, ha='right')
plt.ylim(0, max(income_rates['Attrited']) + 4)
plt.tight_layout()
plt.show()
```

**Answer and analysis:** Attrition rates differ moderately across income categories, ranging from approximately **13.5%** for `$60K - $80K` customers to **17.3%** for `$120K +` customers. The `<$40K` group has a similar rate of **17.2%**. The total spread is only about **3.8 percentage points**, and there is no consistent increase or decrease across the income scale. Thus, income category alone does not show a strong or simple relationship with attrition, and the expectation that lower-income customers would consistently leave more often is not supported.

### Question 3: Are customers who are more inactive more likely to leave the bank?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
if 'df' not in globals():
    df = pd.read_csv('BankChurners.csv')

inactive_rates = (df.assign(Attrited=df['Attrition_Flag'].eq('Attrited Customer'))
                  .groupby('Months_Inactive_12_mon')['Attrited'].agg(['mean', 'size']).reset_index())
inactive_rates['Rate'] = inactive_rates['mean'] * 100
plt.figure(figsize=(9, 5))
ax = sns.barplot(data=inactive_rates, x='Months_Inactive_12_mon', y='Rate', color='#e07a5f')
for i, row in inactive_rates.iterrows():
    ax.text(i, row['Rate'] + 1, f"{row['Rate']:.1f}%\nn={int(row['size'])}", ha='center', fontsize=9)
plt.title('Question 3: Attrition generally rises with inactive months')
plt.xlabel('Months inactive in the last 12 months')
plt.ylabel('Attrition rate (%)')
plt.ylim(0, inactive_rates['Rate'].max() + 10)
plt.tight_layout()
plt.show()
```

**Answer and analysis:** Among the well-represented groups from **1 to 4 inactive months**, attrition rises clearly: about **4.5%**, **15.4%**, **21.5%**, and **29.9%**, respectively. This supports the expectation that greater inactivity is associated with a higher probability of leaving. However, the pattern is not perfectly monotonic at 5 and 6 months. The 0-month group has a very high rate (**51.7%**) but contains only **29 customers**, so it should not be used as strong evidence. Overall, inactivity appears more informative for attrition than income, while the smallest groups require cautious interpretation.

### Question 4: Do existing and attrited customers differ in their transaction activity?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
if 'df' not in globals():
    df = pd.read_csv('BankChurners.csv')

activity_means = df.groupby('Attrition_Flag', observed=False)[['Total_Trans_Amt', 'Total_Trans_Ct']].mean().reindex(['Existing Customer', 'Attrited Customer']).reset_index()
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
sns.barplot(data=activity_means, x='Attrition_Flag', y='Total_Trans_Amt', hue='Attrition_Flag', legend=False, palette=['#59a14f', '#e15759'], ax=axes[0])
axes[0].set_title('Mean annual transaction amount')
axes[0].set_xlabel('')
axes[0].set_ylabel('Amount')
sns.barplot(data=activity_means, x='Attrition_Flag', y='Total_Trans_Ct', hue='Attrition_Flag', legend=False, palette=['#59a14f', '#e15759'], ax=axes[1])
axes[1].set_title('Mean annual transaction count')
axes[1].set_xlabel('')
axes[1].set_ylabel('Count')
fig.suptitle('Question 4: Attrited customers are less active', y=1.02)
plt.tight_layout()
plt.show()
```

**Answer and analysis:** Attrited customers are substantially less active. Their mean transaction count is approximately **44.9**, compared with **68.7** for existing customers, which is about **34.6% lower**. Their mean total transaction amount is approximately **$3,095**, compared with **$4,655**, about **33.5% lower**. The medians show the same pattern: **43 versus 71 transactions** and **$2,329 versus $4,100**. Therefore, the expectation that attrited customers have lower transaction activity is strongly supported by the descriptive results.

### Question 5: Is credit utilization related to customer attrition?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
if 'df' not in globals():
    df = pd.read_csv('BankChurners.csv')

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Attrition_Flag', y='Avg_Utilization_Ratio', hue='Attrition_Flag', legend=False, palette=['#59a14f', '#e15759'], showfliers=False)
plt.title('Question 5: Credit utilization is lower among attrited customers')
plt.xlabel('Customer status')
plt.ylabel('Average utilization ratio')
plt.tight_layout()
plt.show()
```

**Answer and analysis:** Credit utilization differs clearly between the two groups. Attrited customers have a mean utilization ratio of approximately **0.163**, compared with **0.296** for existing customers. The median for attrited customers is **0.000**, while the median for existing customers is **0.211**. In this dataset, customers who left tended to use a smaller share of their available credit. This supports a relationship between utilization and attrition, but the result is associative and does not prove that low utilization causes customers to leave.

### Question 6: Are there substantial differences in attrition between card categories?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
if 'df' not in globals():
    df = pd.read_csv('BankChurners.csv')

card_order = ['Blue', 'Silver', 'Gold', 'Platinum']
card_rates = (df.assign(Attrited=df['Attrition_Flag'].eq('Attrited Customer'))
              .groupby('Card_Category', observed=False)['Attrited'].agg(['mean', 'size']).reindex(card_order).reset_index())
card_rates['Rate'] = card_rates['mean'] * 100
plt.figure(figsize=(8, 5))
ax = sns.barplot(data=card_rates, x='Card_Category', y='Rate', color='#7b6d9d')
for i, row in card_rates.iterrows():
    ax.text(i, row['Rate'] + 1, f"{row['Rate']:.1f}%\nn={int(row['size'])}", ha='center', fontsize=9)
plt.title('Question 6: Attrition rates by card category')
plt.xlabel('Card category')
plt.ylabel('Attrition rate (%)')
plt.ylim(0, card_rates['Rate'].max() + 10)
plt.tight_layout()
plt.show()
```

**Answer and analysis:** Observed attrition rates range from approximately **14.8%** for Silver cards to **25.0%** for Platinum cards. However, the category sizes are extremely unequal: Blue has **9,436 customers**, Silver 555, Gold 116, and Platinum only 20. The Platinum rate is based on only **5 attrited customers**, so it is unstable and should not be treated as evidence of a large category effect. The well-represented Blue and Silver categories differ by only about **1.3 percentage points**. Therefore, the data suggest some observed variation, but they do not provide strong descriptive evidence of substantial differences between card categories.

### Question 7: Are there extreme or unusual values in credit limit, transaction amount, or transaction count?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
if 'df' not in globals():
    df = pd.read_csv('BankChurners.csv')

outlier_columns = ['Credit_Limit', 'Total_Trans_Amt', 'Total_Trans_Ct']
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, column, color in zip(axes, outlier_columns, ['#4c78a8', '#f28e2b', '#76b7b2']):
    q1, q3 = df[column].quantile([0.25, 0.75])
    upper_bound = q3 + 1.5 * (q3 - q1)
    sns.boxplot(y=df[column], color=color, ax=ax)
    ax.axhline(upper_bound, color='crimson', linestyle='--', linewidth=1.4, label=f'IQR upper bound: {upper_bound:,.0f}')
    ax.set_title(column)
    ax.set_xlabel('')
    ax.legend(fontsize=8, loc='upper left')
fig.suptitle('Question 7: High potential outliers under the 1.5 × IQR rule', y=1.03)
plt.tight_layout()
plt.show()
```

**Answer and analysis:** Using the standard **1.5 × IQR rule**, `Credit_Limit` has **984 potential outliers (9.72%)** above approximately **$23,836**, and `Total_Trans_Amt` has **896 potential outliers (8.85%)** above approximately **$8,619**. `Total_Trans_Ct` has only **2 potential outliers (0.02%)** above 135 transactions. These values are unusual relative to the central part of each distribution, but they are still plausible customer records. They should be investigated or handled carefully for specific modeling methods rather than automatically removed from an exploratory analysis.
